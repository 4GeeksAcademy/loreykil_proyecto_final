```python
import pandas as pd
import glob
from pathlib import Path
```

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.options.display.float_format = '{:.2f}'.format

```

```python
# Ruta a la carpeta
data_dir = Path("../data/Export Manta plastics")

# Buscar todos los archivos .tsv
files = sorted(data_dir.glob("*.tsv"))

print(f"Número de archivos encontrados: {len(files)}")

# Leer todos los TSV
dfs = []
for f in files:
    df = pd.read_csv(
        f,
        sep="\t",
        low_memory=False
    )
    df["source_file"] = f.name  # opcional: para trazar origen
    dfs.append(df)

# Unir todo en una sola tabla
df_all = pd.concat(dfs, ignore_index=True)

# Guardar en formato CSV
df_all.to_csv("../data/metadata_img.csv", index=False)

df_all = pd.read_csv("../data/metadata_img.csv")

df_all.shape
```
