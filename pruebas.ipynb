```python
import geopandas as gpd
```

```python
gdf = gpd.read_file("data/Predicciones/mapa_continuo.gpkg")
gdf
```

```python
gdf_profile = gpd.read_file("data/grid_ocean/grilla_agua_perfiles.gpkg")
gdf_profile
```

```python
gdf.info()
```

```python
gdf_profile.info()
```

```python
gdf = gdf.to_crs(gdf_profile.crs)

gdf = gpd.sjoin(
    gdf,
    gdf_profile[["geometry", "profile_eco"]],
    how="left",
    predicate="intersects",
)

gdf = gdf.drop(columns=["index_right"])
gdf
```

```python
gdf.info()
```

```python
gdf.to_file(
    "data/Predicciones/mapa_continuo.gpkg",
    driver="GPKG"
)
```
